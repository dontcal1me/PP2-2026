## 03
- 파일에서 임의의 행을 읽는 프로그램을 작성하시오.


```python
fileName = input("파일 이름을 입력하시오: ")
f = open(f"{fileName}","r", encoding="utf-8" )

readNumber = int(input("행 번호를 입력하세요: "))
mylist = f.read().strip() # 전체를 읽을 필요가 없다~
mylist = mylist.split(".")

print(f"{readNumber}번 행은 {mylist[readNumber - 1]}")


```

```python
# 더 쉬운 방법(클로드)
fileName = input("파일 이름을 입력하시오: ")
n = int(input("줄 번호를 입력하세요: "))

with open(fileName, "r", encoding="utf-8") as f:
    lines = f.readlines()

print(f"{n}번 줄은 {lines[n - 1].strip()}")
```

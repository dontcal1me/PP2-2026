 - 파일은 보조기억장치(영구저장장치) 에 기록된 논리적인 정보 단위이다.

## 작업폴더 : 파이썬 파일을 실행하는 위치
- 가변길이 레코드(한 라인,한 줄) : 예시 이름 여러자 입력가능 3자리 2자리 가 존재함


```python
with open("file.txt", "r", encoding="utf-8") as f:
    #for line in f:
        #print(line.strip())

    line = f.readline().strip()
    while line != "":
        print(line)
        line = f.readline().strip()

    eline = f.readline().strip()
    print(eline)
    print(f.tell()) # 커서의 위치를 출력

    f. seek(0, 0) # 커서를 파일의 처음으로 이동
    eline = f.readline().strip()
    print(eline)
    print(f.tell()) 
```

## 파일 읽고 쓰기

| 파일 모드 | 모드 이름 | 설명 |
| --- | --- | --- |
| "r" | 읽기 모드 | 파일을 처음부터 읽는다 |
| "w" | 쓰기 모드 | 파일의 처음부터 씀, 없으면 생성,만약 존재하면 기존내용 지워지고 새로 작성 |
| "a" | 추가 모드 | 파일의 끝에 쓴다. 없으면 생성 |
| "r+" | 읽기 쓰기 모드 | 읽기와 쓰기 모드 |

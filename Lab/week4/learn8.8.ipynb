# counter 클래스 만들기
- 데이터: 현재 숫자
- 동작: 증가하기, 현재 값 읽어오기

```python
class Counter : #클래스 이름은 항상 대문자로 시작, 항상 명사형
    def __init__(self, count = 0): # 생성 함수  매게변수 self 의 뜻은 자기 자신이라는 뜻
        self.count = count # 초기값 설정

    def increment(self, amount=1): # 매개변수 amount 기본값 1
        self.count += amount 

    def get_count(self): 
        return self.count 

def test_counter():
    myc = Counter() # 객체 생성
    print(myc.get_count())
    myc.increment() # 1 증가
    print(myc.get_count())
    myc.increment() # 1 증가
    print(myc.get_count())
    for _ in range(10): # 10번 반복
        myc.increment() # 1 증가
    print(myc.get_count()) # 12 출력



    yourc = Counter(5)
    print(yourc.get_count())
    yourc.increment()
    print(yourc.get_count())



if __name__ == "__main__":
    test_counter() # 테스트 함수 호출
    
```

# TV 클래스 추가
- 데이터: 현재 채널번호, 볼륨량, ON 여부
- 동작: 켜기, 끄기, 채널변경, 볼륨변경, 현재 TV 객체 내용 출력

- 이름: MyTelevision


```python
class myTelevision:
    serialNumber = 0
    def __init__(self, channel=1, volume=10, is_on=False):
        self.channel = channel
        self.volume = volume
        self.is_on = is_on
        myTelevision.serialNumber += 1
        self.myserial = myTelevision.serialNumber
        
    def turn_on(self):
        self.is_on = True

    def turn_off(self):
        self.is_on = False

    def change_channel(self, channel):
        self.channel = channel

    def change_volume(self, volume):
        self.volume = volume

    def display_status(self):
        print(f"Channel: {self.channel}, Volume: {self.volume}, is_on: {'On' if self.is_on else 'Off'}")



def test_myTelevision():
    mytv = myTelevision(channel = 11)  # 객체 생성
    mytv.turn_on()
    mytv.display_status()
    mytv.turn_off()
    mytv.display_status()

if __name__ == "__main__":
    test_myTelevision()  # 테스트 함수 호출
```

# TV 리모콘 클래스

- 데이터는 없고 동작만 있음
- 연관된 티비를 켜거나 끄라는 동작만 지원


```python



class TvRemocon:
    def __init__(self, tv: myTelevision):
        self.tv = tv

    def press_on(self):
        self.tv.turn_on()

    def press_off(self):
        self.tv.turn_off()

    def press_display(self):
        self.tv.display_status()


def test_remocon():
    tv1 = myTelevision(channel=7, volume=15)
    rc1 = TvRemocon(tv1)

    rc1.press_on()
    rc1.press_display()   # 켜진 상태 표시

    rc1.press_off()
    rc1.press_display()   # 꺼진 상태 표시


if __name__ == "__main__":
    test_remocon()
```

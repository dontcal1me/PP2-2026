### 1. 튜플 (Tuple)
특징: 순서가 있고, 한 번 생성하면 값을 수정·추가·삭제할 수 없는(Immutable) 자료형입니다.
주요 용도: 데이터의 변경을 막고 안전하게 보호하고 싶을 때 사용합니다. 소괄호 ()로 생성합니다.

```python
# 튜플 생성
point = (10, 20)
user_info = ("Alice", 25, "Developer")

# 인덱싱 (순서가 있으므로 위치로 접근 가능)
print(point[0])  # 10
print(user_info[2])  # Developer

# 언패킹 (여러 변수에 한 번에 값 할당)
x, y = point
print(f"X좌표: {x}, Y좌표: {y}")

# 튜플의 값은 변경 불가 (오류 발생)
# point[0] = 30  -> TypeError 발생!
```

2. 집합 (Set)
특징: 순서가 없고, 중복을 허용하지 않는 자료형입니다.

주요 용도: 중복 데이터 제거, 합집합·교집합·차집합 등 집합 연산이 필요할 때 사용합니다. 중괄호 {}로 생성합니다.

```python
# 집합 생성 (중복된 값은 자동으로 제거됨)
fruits = {"apple", "banana", "cherry", "apple"}
print(fruits)  # {'apple', 'banana', 'cherry'}

# 요소 추가 및 삭제
fruits.add("orange")
fruits.remove("banana")

# 리스트의 중복 값 제거에 활용
numbers = [1, 2, 2, 3, 4, 4, 4, 5]
unique_numbers = set(numbers)
print(unique_numbers)  # {1, 2, 3, 4, 5}

# 집합 연산 (합집합, 교집합, 차집합)
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}

print(set1 | set2)  # 합집합: {1, 2, 3, 4, 5, 6}
print(set1 & set2)  # 교집합: {3, 4}
print(set1 - set2)  # 차집합: {1, 2}
```

3. 딕셔너리 (Dictionary)
특징: 키(Key)와 값(Value)의 쌍으로 데이터를 저장하는 구조입니다.

주요 용도: 연관된 데이터를 이름(Key)으로 빠르게 찾고 관리할 때 사용합니다. 중괄호 { Key: Value }로 생성합니다.

```python
# 딕셔너리 생성
student = {"name": "김철수", "age": 20, "major": "컴퓨터공학"}

# 값 조회 (Key 사용)
print(student["name"])  # 김철수
print(student.get("age"))  # 20

# 값 수정 및 추가
student["age"] = 21  # 기존 값 수정
student["grade"] = "A"  # 새로운 Key-Value 쌍 추가
print(student)

# Key, Value 전체 순회
for key, value in student.items():
    print(f"{key}: {value}")
```

### byte, bytes, bytearray 

#### bytes
생성 후 값 변경 불가
문자열 앞에 b를 붙이거나 bytes() 함수로 생성합니다.

#### bytearray
생성 후 값 수정·추가 가능(list와 비슷함)
내부 요소를 직접 수정하거나 append() 등의 메서드로 데이터를 늘릴 수 있습니다


```python
# 생성: 문자열 앞에 b 접두사 붙이기
data = b"Hello"

print(data[0])  # Output: 72 ('H'의 ASCII 정수 값)
print(data)     # Output: b'Hello'

# 수정 시도 시 에러 발생
data[0] = 65  
#-> TypeError: 'bytes' object does not support item assignment
```

```python
# bytes를 기반으로 bytearray 생성
mutable_data = bytearray(b"Hello")

# 첫 번째 글자 'H'(72)를 'J'(74)로 변경
mutable_data[0] = 74
print(mutable_data)  # Output: bytearray(b'Jello')

# 데이터 추가
mutable_data.append(33)  # 33은 ASCII '!'
print(mutable_data)  # Output: bytearray(b'Jello!')
```

### 세트의 내장함수 all(), any()

#### all()
모든 요소가 참(True)이어야 True를 반환합니다. 하나라도 거짓(False)이면 False입니다. (AND 연산과 유사)

all() 함수는 요소끼리 값이 "같다/다르다"를 비교하는 함수가 아닙니다.
all()은 들어있는 요소들이 "참(True)인지 거짓(False)인지" 판단하는 함수입니다. 모든 요소가 참(True)에 해당할 때만 True를 반환합니다.

#### any()
요소 중 단 하나라도 참(True)이면 True를 반환합니다. 모두 거짓일 때만 False입니다. (OR 연산과 유사)




```python
# 기본 동작 예시 (Boolean 데이터)
bool_set_1 = {True, True, True}
bool_set_2 = {True, False, True}

print(all(bool_set_1))  # True (모두 True이므로)
print(all(bool_set_2))  # False (False가 1개 섞여있으므로)

print(any(bool_set_2))  # True (True가 최소 1개 있으므로)
print(any({0, "", False}))  # False (모두 False 취급되는 값들)
```

```python
# 1, 2, "hello"는 모두 서로 다른 값이지만, 파이썬에서 모두 '참(True)' 취급을 받습니다.
data = {1, 2, "hello"}

print(all(data))  # 출력: True

# 모든 요소가 0으로 '같지만', 0은 파이썬에서 '거짓(False)' 취급을 받습니다.
data = [0, 0, 0]

print(all(data))  # 출력: False
```

### byte, bytes, bytearray 

#### bytes
생성 후 값 변경 불가
문자열 앞에 b를 붙이거나 bytes() 함수로 생성합니다.

#### bytearray
생성 후 값 수정·추가 가능(list와 비슷함)
내부 요소를 직접 수정하거나 append() 등의 메서드로 데이터를 늘릴 수 있습니다


```python
# 생성: 문자열 앞에 b 접두사 붙이기
data = b"Hello"

print(data[0])  # Output: 72 ('H'의 ASCII 정수 값)
print(data)     # Output: b'Hello'

# 수정 시도 시 에러 발생
data[0] = 65  
#-> TypeError: 'bytes' object does not support item assignment
```

```python
# bytes를 기반으로 bytearray 생성
mutable_data = bytearray(b"Hello")

# 첫 번째 글자 'H'(72)를 'J'(74)로 변경
mutable_data[0] = 74
print(mutable_data)  # Output: bytearray(b'Jello')

# 데이터 추가
mutable_data.append(33)  # 33은 ASCII '!'
print(mutable_data)  # Output: bytearray(b'Jello!')
```

### 세트의 내장함수 all(), any()

#### all()
모든 요소가 참(True)이어야 True를 반환합니다. 하나라도 거짓(False)이면 False입니다. (AND 연산과 유사)

all() 함수는 요소끼리 값이 "같다/다르다"를 비교하는 함수가 아닙니다.
all()은 들어있는 요소들이 "참(True)인지 거짓(False)인지" 판단하는 함수입니다. 모든 요소가 참(True)에 해당할 때만 True를 반환합니다.

#### any()
요소 중 단 하나라도 참(True)이면 True를 반환합니다. 모두 거짓일 때만 False입니다. (OR 연산과 유사)




```python
# 기본 동작 예시 (Boolean 데이터)
bool_set_1 = {True, True, True}
bool_set_2 = {True, False, True}

print(all(bool_set_1))  # True (모두 True이므로)
print(all(bool_set_2))  # False (False가 1개 섞여있으므로)

print(any(bool_set_2))  # True (True가 최소 1개 있으므로)
print(any({0, "", False}))  # False (모두 False 취급되는 값들)
```

```python
# 1, 2, "hello"는 모두 서로 다른 값이지만, 파이썬에서 모두 '참(True)' 취급을 받습니다.
data = {1, 2, "hello"}

print(all(data))  # 출력: True

# 모든 요소가 0으로 '같지만', 0은 파이썬에서 '거짓(False)' 취급을 받습니다.
data = [0, 0, 0]

print(all(data))  # 출력: False
```

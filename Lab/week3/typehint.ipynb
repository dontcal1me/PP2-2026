이 파일은 파이썬의 typehint 예제와 이론을 실습한 결과입니다. 작성자 202611831 윤종하

타입힌트는 함수의 매개변수와 반환값의 타입을 지정하는 목적으로 사용하며, 엄격한 타입지정으로 인한 프로그래머의 의도를 정확하게 전달하기 위함이다. 안써도 되지만 파이썬 13부터는 쓰는것을 권장하고 있다.

```python
def greet(mas: str) -> str:
    print("in greet: ", mas)
    return mas +"!"

print(greet("Hello"))
```

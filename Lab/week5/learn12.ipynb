## 1. 상속(Inheritance)
- 기존에 존재하는 클래스로부터 **코드와 데이터를 이어받고**, 자신이 필요한 기능을 추가하는 기법
- 상속에서는 자식 클래스에서 부모 클래스로 화살표를 그린다. (UML)
- 예: `Vehicle` ← `Car`, `Truck`, `Bus` / `Car` ← `SportsCar`

### 왜 상속을 사용하는가?
- `Car`, `Truck`, `Bus`에 `setSpeed()`, `turn()` 같은 코드가 **중복**된다.
- 중복되는 코드를 부모 클래스(`Vehicle`)에 모으면 코드 재사용·유지보수가 쉬워진다.

### 상속과 is-a 관계
상속은 클래스 간의 **"is-a"** 관계를 만드는 데 사용한다. (푸들은 강아지이다, 자동차는 차량이다, 꽃은 식물이다, 사각형은 모양이다)

| 부모 클래스 | 자식 클래스 |
|-------------|-------------|
| Animal(동물) | Lion, Dog, Cat |
| Bike(자전거) | MountainBike, RoadBike, TandemBike |
| Vehicle(탈것) | Car, Bus, Truck, Boat, Motorcycle, Bicycle |
| Student(학생) | GraduateStudent, UnderGraduate |
| Person(사람) | Student, Employee |
| Shape(도형) | Rectangle, Triangle, Circle |

### 상속 구현하기 (문법)

    def 메소드1(self, ...):

### 예제: Car → ElectricCar


````python
class Car:
    """일반적인 자동차를 나타내는 클래스"""
    def __init__(self, make, model, color, price):
        self.make = make        # 메이커
        self.model = model      # 모델
        self.color = color      # 자동차의 색상
        self.price = price      # 자동차의 가격

    def setMake(self, make):    # 설정자 메소드
        self.make = make

    def getMake(self):          # 접근자 메소드
        return self.make

    # 차량에 대한 정보를 문자열로 요약하여 반환한다.
    def getDesc(self):
        return "차량 =(" + str(self.make) + "," + \
               str(self.model) + "," + \
               str(self.color) + "," + \
               str(self.price) + ")"


class ElectricCar(Car):                                 # 부모 클래스 Car 상속
    def __init__(self, make, model, color, price, batterySize):
        super().__init__(make, model, color, price)     # 부모 생성자 호출
        self.batterySize = batterySize

    def setBatterySize(self, batterySize):              # 설정자 메소드
        self.batterySize = batterySize

    def getBatterySize(self):                           # 접근자 메소드
        return self.batterySize


def main():
    myCar = ElectricCar("Tisla", "Model S", "white", 10000, 0)
    myCar.setMake("Tesla")        # 부모에게 물려받은 설정자 메소드 호출
    myCar.setBatterySize(60)      # 자식 클래스의 설정자 메소드 호출
    print(myCar.getDesc())        # 전기차 객체를 문자열로 출력
    print("배터리:", myCar.getBatterySize())

main()
````

## 2. 부모 클래스의 생성자 호출 – `super()`
- 부모 클래스 = **Super-class**, 자식 클래스 = **Sub-class**
- `super()`는 자식 클래스에서 **부모 클래스를 참조**하는 함수
- 자식 클래스는 부모 클래스의 **모든 것(속성, 연산)** 을 상속받는다.
  - 부모의 **private** 속성은 상속은 하되 접근할 수 없다.
  - 부모의 **protected** 속성은 접근할 수 있다.
  - **public** 은 모든 클래스가 접근할 수 있다.
- 자식 `__init__`에서 `super().__init__(...)`으로 **부모 생성자를 호출하지 않으면** 부모의 변수가 만들어지지 않아 오류가 난다.

````python
class Animal:
    def __init__(self, age=0):
        self.age = age

    def eat(self):
        print("동물이 먹고 있습니다.")


# 부모 클래스의 생성자를 호출하지 않았다!
class Dog(Animal):
    def __init__(self, age=0, name=""):
        self.name = name


d = Dog()
try:
    print(d.age)            # 부모 생성자가 호출되지 않아 age 변수가 생성되지 않았다.
except AttributeError as e:
    print("AttributeError:", e)


# 올바른 방법
class Dog2(Animal):
    def __init__(self, age=0, name=""):
        super().__init__(age)       # 부모 생성자 호출
        self.name = name


d2 = Dog2(3, "바둑이")
print(d2.age, d2.name)
d2.eat()                            # 부모 메소드도 그대로 사용
````

## 3. `type()` 과 `isinstance()`
- `type(obj)` : 객체의 **정확한** 클래스를 알려준다.
- `isinstance(obj, 클래스)` : obj가 그 클래스(또는 **자식 클래스**)의 객체이면 `True`

````python
x = Animal()
y = Dog2()
print(type(x))
print(type(y))
print(isinstance(x, Animal), isinstance(y, Animal))   # True True  (Dog2는 Animal이기도 하다)
print(type(y) == Animal)                              # False (type은 정확한 클래스만 비교)
print(issubclass(Dog2, Animal))                       # True
````

## 4. private 멤버와 부모 클래스
- 부모의 `__money`는 private이라 **자식 클래스에서 사용할 수 없다.**
- 이름 앞의 `__`는 내부적으로 `_부모클래스명__money`로 바뀐다(이름 맹글링).

````python
class Parent(object):
    def __init__(self):
        self.__money = 100


class Child(Parent):
    def __init__(self):
        super().__init__()


obj = Child()
try:
    print(obj.money)                 # 오류
except AttributeError as e:
    print("AttributeError:", e)

print(obj.__dict__)                  # 실제 저장된 이름 확인
````

## 5. 다중 상속
- 파이썬은 **다중 상속**을 지원한다.
- 부모가 한 명 이상이므로 `super()`보다는 **부모 클래스의 이름**으로 접근한다. (`Person.__init__(self, ...)`)

````python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def show(self):
        print(self.name, self.age)


class Student:
    def __init__(self, id):
        self.id = id

    def getId(self):
        return self.id


class CollegeStudent(Person, Student):
    def __init__(self, name, age, id):
        Person.__init__(self, name, age)
        Student.__init__(self, id)


obj = CollegeStudent('Kim', 22, '100036')
obj.show()
print(obj.getId())
print(CollegeStudent.__mro__)      # 메소드 탐색 순서
````

## 6. 메소드 오버라이딩(Overriding)
- "자식 클래스의 메소드가 부모 클래스의 메소드를 **오버라이드(재정의)** 한다"고 말한다.
- 부모 메소드를 자식이 **자신의 필요에 맞추어 변경**하는 것
- **다형성**: 함수 이름은 같으나 함수가 속한 클래스에 따라 다른 코드가 수행되는 것 (메소드 오버라이드도 다형성의 일종)
- 부모 메소드까지 함께 실행하려면 `super().메소드()` 를 호출한다.

````python
import math


class Shape:
    def __init__(self):
        pass

    def draw(self):
        print("draw()가 호출됨")

    def get_area(self):
        print("get_area()가 호출됨")


class Circle(Shape):
    def __init__(self, radius=0):
        super().__init__()
        self.radius = radius

    def draw(self):                          # 오버라이딩
        print("원을 그립니다.")

    def get_area(self):                      # 오버라이딩
        return math.pi * self.radius ** 2


c = Circle(10)
c.draw()
print("원의 면적:", c.get_area())


# 부모 메소드를 함께 호출하는 오버라이딩
class Circle2(Shape):
    def __init__(self, radius=0):
        super().__init__()
        self.radius = radius

    def draw(self):
        super().draw()                       # 부모 클래스의 draw()가 호출된다.
        print("원을 그립니다.")


Circle2(5).draw()
````

### Lab 2: 직원과 매니저
직원(Employee)은 월급만, 매니저(Manager)는 월급 외에 보너스가 있다. `Employee.getSalary()`를 `Manager`에서 재정의한다.

(슬라이드의 `return salary`는 `self.salary`로 써야 하므로 바로잡았다.)

````python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def getSalary(self):
        return self.salary


class Manager(Employee):
    def __init__(self, name, salary, bonus):
        super().__init__(name, salary)
        self.bonus = bonus

    def getSalary(self):                       # 오버라이딩
        salary = super().getSalary()
        return salary + self.bonus

    def __repr__(self):
        return "이름: " + self.name + "; 월급: " + str(self.salary) + \
               "; 보너스: " + str(self.bonus)


kim = Manager("김철수", 2000000, 1000000)
print(kim)
print("총 지급액:", kim.getSalary())
````

### Lab 3: 은행 계좌 (BankAccount → SavingsAccount, CheckingAccount)
- `BankAccount`: balance, name, number, `withdraw()`, `deposit()`
- `SavingsAccount`: `interest_rate`, `add_interest()`
- `CheckingAccount`: `withdraw_charge`(수표 1회 발행 수수료), `withdraw()` = 찾을 금액 + 수수료

````python
class BankAccount:
    def __init__(self, name, number, balance):
        self.balance = balance
        self.name = name
        self.number = number

    def withdraw(self, amount):
        self.balance -= amount
        return self.balance

    def deposit(self, amount):
        self.balance += amount
        return self.balance


class SavingsAccount(BankAccount):
    def __init__(self, name, number, balance, interest_rate):
        super().__init__(name, number, balance)
        self.interest_rate = interest_rate

    def set_interest_rate(self, interest_rate):
        self.interest_rate = interest_rate

    def get_interest_rate(self):
        return self.interest_rate

    def add_interest(self):                       # 예금에 이자를 더한다.
        self.balance += self.balance * self.interest_rate


class CheckingAccount(BankAccount):
    def __init__(self, name, number, balance):
        super().__init__(name, number, balance)
        self.withdraw_charge = 10000              # 수표 발행 수수료

    def withdraw(self, amount):                   # 오버라이딩
        return BankAccount.withdraw(self, amount + self.withdraw_charge)


a1 = SavingsAccount("홍길동", 123456, 10000, 0.05)
a1.add_interest()
print("저축예금의 잔액=", a1.balance)

a2 = CheckingAccount("김철수", 123457, 2000000)
a2.withdraw(100000)
print("당좌예금의 잔액=", a2.balance)
````

### Lab 4: Card 와 Deck
`Card` 클래스와 52개의 `Card` 객체를 가지고 있는 `Deck` 클래스를 만들고, 각각 `__str__()`을 구현해 출력한다.

````python
class Card:
    suitNames = ['클럽', '다이아몬드', '하트', '스페이드']
    rankNames = [None, '에이스', '2', '3', '4', '5', '6', '7',
                 '8', '9', '10', '잭', '퀸', '킹']

    def __init__(self, suit, rank):
        self.suit = suit
        self.rank = rank

    def __str__(self):
        return Card.suitNames[self.suit] + " " + \
               Card.rankNames[self.rank]


class Deck:
    def __init__(self):
        self.cards = []

        for suit in range(4):
            for rank in range(1, 14):
                card = Card(suit, rank)
                self.cards.append(card)

    def __str__(self):
        lst = [str(card) for card in self.cards]
        return str(lst)


deck = Deck()          # 덱 객체를 생성한다.
print(deck)            # 덱 객체를 출력한다. __str__()이 호출된다.
print(len(deck.cards))
````

## 8. 다형성(Polymorphism)
- **많은(poly) + 모양(morph)** : 하나의 식별자로 **다양한 타입(클래스)** 을 처리하는 것
- 동일한 코드로 다양한 타입의 객체를 처리할 수 있는 기법
- 도형의 타입에 상관없이 그리려면 무조건 `draw()`, 면적을 구하려면 무조건 `getArea()`를 호출하면 된다.
- 내장 함수도 다형성을 가진다. 예) `len()`은 리스트, 문자열, 딕셔너리 모두에 사용 가능

````python
class Shape:
    def __init__(self, name):
        self.name = name

    def getArea(self):
        raise NotImplementedError("이것은 추상메소드입니다.")


class Circle(Shape):
    def __init__(self, name, radius):
        super().__init__(name)
        self.radius = radius

    def getArea(self):
        return 3.141592 * self.radius ** 2


class Rectangle(Shape):
    def __init__(self, name, width, height):
        super().__init__(name)
        self.width = width
        self.height = height

    def getArea(self):
        return self.width * self.height


shapeList = [Circle("c1", 10), Rectangle("r1", 10, 10)]
for s in shapeList:
    print(s.getArea())          # 객체의 타입에 따라 다른 getArea()가 호출된다(동적 바인딩)

# 내장 함수와 다형성
mylist = [1, 2, 3]
print("리스트의 길이=", len(mylist))
s = "This is a sentense"
print("문자열의 길이=", len(s))
d = {'aaa': 1, 'bbb': 2}
print("딕셔너리의 길이=", len(d))
````

## 9. `object` 클래스
- 모든 클래스의 맨 위에는 `object` 클래스가 있다고 생각하면 된다. (일반 클래스 → 특수 클래스로 내려감)
- `class A:` 와 `class A(object):` 는 같다.

| 메소드 | 설명 |
|--------|------|
| `__init__(self[, args...])` | 생성자 → `obj = className(args)` |
| `__del__(self)` | 소멸자 → `del obj` |
| `__repr__(self)` | 객체 표현 문자열 반환 → `repr(obj)` |
| `__str__(self)` | 문자열 표현 반환 → `str(obj)` |
| `__cmp__(self, x)` | 객체 비교 → `cmp(obj, x)` (**파이썬 2** 방식. 파이썬 3에서는 `__eq__`, `__lt__`, `__gt__` 등을 사용) |

````python
class Book:
    def __init__(self, title, isbn):
        self.__title = title
        self.__isbn = isbn

    def __repr__(self):
        return "ISBN: " + self.__isbn + "; TITLE: " + self.__title


book = Book("The Python Tutorial", "0123456")
print(book)


class MyTime:
    def __init__(self, hour, minute, second=0):
        self.hour = hour
        self.minute = minute
        self.second = second

    def __str__(self):
        return '%.2d:%.2d:%.2d' % (self.hour, self.minute, self.second)


time = MyTime(10, 25)
print(time)
print(issubclass(Book, object), isinstance(book, object))
````

## 10. 클래스 간의 관계
- **is-a 관계 (상속)** : 승용차는 차량의 일종이다(Car is a Vehicle), 강아지는 동물의 일종이다, 원은 도형의 일종이다. (UML: `Derived` —extends→ `Base`)
- **has-a 관계 (구성)** : 도서관은 책을 가지고 있다(Library has a book), 거실은 소파를 가지고 있다.
  - 구성(Composition, ◆): 부분이 전체에 종속 (Car ◆ Carburetor)
  - 집합(Aggregation, ◇): 부분이 독립적으로 존재 (Pond ◇ Duck)
- `MyCar` **is-a** `Car`, `MyCar` **has-a** `Engine`

````python
class Animal(object):
    pass


class Dog(Animal):                 # is-a : Dog is an Animal
    def __init__(self, name):
        self.name = name


class Person(object):
    def __init__(self, name):
        self.name = name
        self.pet = None            # has-a : Person has a pet


dog1 = Dog("dog1")
person1 = Person("홍길동")
person1.pet = dog1
print(person1.name, "의 애완동물:", person1.pet.name)
print(isinstance(person1.pet, Animal))
````

# 가시성: Public, protected vate

- 파이썬은 C++/자바와 달리 접근 제어자(public, protected, private)가 **엄격하게 적용되지 않는다.** 하지만 **관례적으로** 다음과 같이 가시성을 표현한다.

| 구분 | 표기 | 의미 |
|------|------|------|
| public | 이름 앞에 밑줄 없음 (`value`) | 어디서나 접근 가능 |
| protected | 이름 앞에 밑줄 1개 (`_value`) | 하위 클래스에서 접근 권장, 외부에서 직접 접근은 권장하지 않음 |
| private | 이름 앞에 밑줄 2개 (`__value`) | 클래스 내부에서만 접근. **이름 맹글링**되어 외부/하위 클래스에서 직접 접근 어렵게 만듦 |

````python
class Base:
    def __init__(self):
        self.public_value = 1           # public
        self._protected_value = 2       # protected
        self.__private_value = 3        # private


class Child(Base):
    def show(self):
        print(self.public_value)        # 가능
        print(self._protected_value)    # 가능
        try:
            print(self.__private_value) # 불가 (AttributeError)
        except AttributeError as e:
            print("AttributeError:", e)


class Other:
    def show(self, obj: Base):
        print(obj.public_value)         # 가능
        print(obj._protected_value)     # 가능 (하지만 권장되지 않음)
        try:
            print(obj.__private_value)  # 불가 (AttributeError)
        except AttributeError as e:
            print("AttributeError:", e)


Child().show()
print("---")
Other().show(Base())
print(Base()._Base__private_value)      # 맹글링된 이름으로는 접근 가능(권장 X)
````

## 11. 추상 메소드(abstract method)
- `abc` 모듈의 `ABC`(Abstract Base Class)와 `@abstractmethod` 데코레이터를 사용해 정의한다.
- 추상 메소드는 **반드시 하위 클래스에서 구현**해야 하며, 추상 메소드를 하나라도 가진 클래스는 **추상 클래스**로서 **직접 인스턴스를 만들 수 없다.**
- 추상 클래스는 `abc.ABC` 클래스로부터 상속받아야 하고, 추상 메소드는 `@abstractmethod`를 선언한다. (보통 `pass`)

````python
from abc import ABC, abstractmethod


class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass


class Dog(Animal):
    def sound(self):
        return "멍멍"


try:
    a = Animal()            # 에러: 추상 클래스는 인스턴스화 불가
except TypeError as e:
    print("TypeError:", e)

d = Dog()
print(d.sound())            # "멍멍"
````

## 12. 정적 멤버(static variable, static method)
- 파이썬의 static member(정적 멤버, **클래스 변수**)는 **클래스 전체에서 공유**되는 변수 (클래스 변수, 클래스 메소드라고 함)
- 메소드는 `@staticmethod` 데코레이터로 정적 메소드를 정의
- 정적 변수는 클래스 안에 `self`가 아닌 일반 변수로 선언하고 `클래스.변수` 형태로 접근

````python
class Counter:
    count = 0                         # 클래스 변수(정적 멤버)

    def __init__(self):
        self.__object_counter = 1     # private 인스턴스 변수

    @staticmethod
    def increment_class():
        Counter.count += 1

    def increment_object(self):
        self.__object_counter += 1

    def get_object_counter(self):
        return self.__object_counter


Counter.increment_class()
print(Counter.count)                  # 1

c = Counter()                         # 인스턴스 생성
print(c.get_object_counter())         # 1
c.increment_object()
print(c.get_object_counter())         # 2
try:
    print(c.__object_counter)         # private 접근 불가
except AttributeError as e:
    print("AttributeError:", e)
````

### 연습문제 (추상클래스 다루기)
- 동물(`Animal`)을 울음을 **추상 메소드**로 하는 추상클래스로 정의하고, 개(`Dog`)와 고양이(`Cat`)를 구체적인 클래스로 정의한 뒤 울음소리를 출력한다.
- 각 동물 객체는 자신의 이름을 속성으로 가지며 생성자에서 이름을 초기화한다.
- `Animal`은 현재 생성된 동물 객체의 수를 **클래스 변수**로 가지고, **정적 메소드**로 개수를 반환한다. 객체가 생성(`__init__`)될 때 증가, 삭제(`__del__`)될 때 감소한다.
- 개/고양이 객체들을 리스트로 만들어 `Animal` 변수로 받아 **동적 바인딩**을 확인하고, 모든 객체가 울게 하는 함수를 작성해 테스트한다.

````python
from abc import ABC, abstractmethod


class Animal(ABC):
    animal_count = 0                       # 클래스 변수

    def __init__(self, name):
        self.name = name
        Animal.animal_count += 1           # 생성 시 증가

    def __del__(self):
        Animal.animal_count -= 1           # 삭제 시 감소

    @abstractmethod
    def sound(self):
        pass

    @staticmethod
    def get_animal_count():
        return Animal.animal_count


class Dog(Animal):
    def sound(self):
        return "멍멍"


class Cat(Animal):
    def sound(self):
        return "야옹"


def animal_sound(animal: Animal):
    print(f"{animal.name}: {animal.sound()}")


def animal_sounds(animals):
    for animal in animals:
        animal_sound(animal)


if __name__ == "__main__":
    dog: Dog = Dog("바둑이")
    print(dog.sound())                      # 멍멍

    cat = Cat("야옹이")
    print(cat.sound())                      # 야옹

    pet: Animal = Dog("바둑이2")
    print(pet.sound())                      # 멍멍 (동적 바인딩)
    pet: Animal = Cat("야옹이2")
    print(pet.sound())                      # 야옹 (동적 바인딩)

    alist: list[Animal] = [Dog("바둑이a"), Cat("나비b"), Dog("멍멍이c"), Cat("야옹이d")]

    animal_sound(alist[2])
    animal_sounds(alist)

    print(f"총 동물의 수: {Animal.get_animal_count()}")

    # 올바른 객체 삭제 방법
    del dog        # del 키워드 사용
    del cat        # del 키워드 사용

    print(f"총 동물의 수: {Animal.get_animal_count()}")
````

> **참고:** 슬라이드 주석에는 총 동물 수가 8 → 6으로 적혀 있지만, 실제로 실행하면 7 → 5가 출력된다. `pet`에 다른 객체(`Cat("야옹이2")`)를 다시 대입하면 이전 객체(`바둑이2`)를 가리키는 변수가 없어져 자동으로 삭제(`__del__`)되기 때문이다. 즉 객체의 수 = 현재 살아 있는 객체의 수이다.

## 13. Dunder methods
- **D**ouble **Under**lines 의 약자
- 파이썬에서 언더라인은 특별한 의미로 사용할 때가 많다. `__XXX__` 메소드는 **자동으로 호출**되는 경우가 많다.
- 이러한 메소드를 **재정의**하여 사용자 코드를 수행하도록 할 수 있다. (`dir(object)`로 확인)

| 메소드 | 설명 | 호출 예 |
|--------|------|---------|
| `__init__(self, ...)` | 클래스 생성 함수, 객체가 만들어질 때 자동 실행 | `a = A()` |
| `__str__(self)` | 클래스 객체를 문자열로 형변환 할 때 자동 실행 | `print(a)` |
| `__eq__(self, other)` | 이 객체와 다른 객체가 같은지 여부 | `a == b` |
| `__gt__(self, other)` | 이 객체가 다른 객체보다 큰지 여부 | `a > b` |

````python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"Student(name={self.name}, age={self.age})"

    def __eq__(self, other):
        if isinstance(other, Student):
            return self.name == other.name and self.age == other.age
        return False

    def __del__(self):
        print(f"Student {self.name} is being deleted")


if __name__ == "__main__":
    student1 = Student("Alice", 20)
    student2 = Student("Bob", 22)
    student3 = Student("Alice", 20)

    print(student1)                  # Calls __str__
    print(student2)                  # Calls __str__

    print(student1 == student2)      # Calls __eq__
    print(student1 == student3)      # Calls __eq__

    del student1                     # Calls __del__
````

## 14. 파이썬 데코레이터(decorator)
- 함수나 클래스의 **동작을 수정하거나 확장**할 때 사용하는 문법
- 함수(또는 메소드) 정의 위에 `@데코레이터이름` 형태로 사용
- 해당 함수가 호출될 때 **데코레이터 함수 안에서 해당 함수가 실행**된다.
- 데코레이터 함수는 `wrapper`라는 내부 함수를 정의해서 반환하는 형식

```python
def deco_func(func):
    def wrapper(*args, **kwargs):
        ...
        result = func(*args, **kwargs)
        ...
        return result
    return wrapper
```

### 기존 함수를 확장하는 2가지 방법
1. **함수 재할당** : 장식된 함수명을 기존 함수명 그대로 사용 (기존 클라이언트 코드 변경 없음) → `원본함수명 = decorator(원본함수명)`
2. **다른 함수명** : 장식된 함수를 새 이름으로 사용 → `신규함수명 = decorator(원본함수명)`

(슬라이드 예제의 `Def`, `If`는 오타 → `def`, `if`)

````python
def simple_decorator(func):
    def wrapper():
        print("함수 실행 전")       # 추가적인 코드
        result = func()            # 목적 함수 실행
        print("함수 실행 후")       # 추가적인 코드
        return result
    return wrapper


@simple_decorator
def say_hello():
    print("Hello, World!")


if __name__ == "__main__":
    print("=== 예제 1: 기본 데코레이터 ===")
    say_hello()


# 방법 2: 데코레이터 문법(@) 없이 직접 적용
def say_hi():
    print("Hello, World!")

original_hi = say_hi                                   # 원본 백업
say_hi = simple_decorator(original_hi)                 # 1. 함수 재할당
decorated_say_hi = simple_decorator(original_hi)       # 2. 다른 함수명

if __name__ == "__main__":
    print("=== 예제 2 ===")
    say_hi()
    decorated_say_hi()
````

## 15. 수업중 작성한 실습 (Car → ElectricCar / Truck)
- `Car`: maker, model, color, price + setter/getter, `getDesc`, `__str__`
- `ElectricCar(Car)`: `batterySize` 속성 추가, `__str__` 
- `Truck(Car)`: `payLoad` 속성 추가, `__str__`
- `isinstance(truck1, Truck)`, `isinstance(truck1, Car)` 모두 `True` → 자식 객체는 부모 타입이기도 하다

````python
class Car:
    def __init__(self, maker, model, color, price):
        self.maker = maker
        self.model = model
        self.color = color
        self.price = price

    def setMaker(slef, maker):
        slef.maker = maker

    def getMaker(self):
        return self.maker

    def getDesc(self):
        return f"Maker: {self.maker}, Model: {self.model}, Color: {self.color}, Price: {self.price}"

    def __str__(self):
        return f"Maker: {self.maker}, Model: {self.model}, Color: {self.color}, Price: {self.price}"

class ElectricCar(Car):
    def __init__(self, maker, model, color, price, batterySize):
        super().__init__(maker, model, color, price)
        self.batterySize = batterySize

    def getBatterySize(self):
        return self.batterySize

    def setBatterySize(self, batterySize):
        self.batterySize = self.batterySize

    def __str__(self):
        return f"{super().__str__()}, bettery: {self.batterySize}KWh"

    def getDesc(self):
        return super().getDesc()

class Truck(Car):
    def __init__(self, maker, model, color, price, payLoad):
        super().__init__(maker, model, color, price)
        self.payLoad = payLoad

    def __str__(self):
        return f"{super().__str__()}, PayLoad: {self.payLoad}"

car1 = Car("Kia", "Ray", "Bule", 2000)
print(car1.getDesc())
print(car1.model)

ecar1 = ElectricCar("Tesla", "Modek S", "Red", 8000, 81)
print(ecar1)
truck1 = Truck("Ford", "F-150", "Blue", 8000, "류지언")
print(truck1)

print(isinstance(truck1, Truck))
print(isinstance(truck1, Car))

````

````python
# issubclass 와 MRO(메소드 탐색 순서) 확인
print(issubclass(ElectricCar, Car), issubclass(Truck, Car), issubclass(Car, Truck))
print(Truck.__mro__)

# 오버라이딩하지 않은 메소드는 부모 것을 그대로 사용
print(truck1.getDesc())
````

## 이번 장에서 배운 것
- 상속은 다른 클래스를 **재사용**하는 탁월한 방법이다. 객체와 객체 간의 **is-a** 관계가 성립된다면 상속을 이용하도록 한다.
- 상속을 사용하면 **중복된 코드**를 줄일 수 있다. 공통적인 코드는 부모 클래스를 작성하여 한 곳으로 모은다.
- 상속에서는 부모 클래스의 메소드를 자식 클래스가 재정의할 수 있다. 이것을 **메소드 오버라이딩**이라고 한다.
- `super()`로 부모 생성자/메소드를 호출하고, 다중 상속에서는 부모 클래스 이름으로 호출한다.
- 가시성: `name`(public), `_name`(protected, 관례), `__name`(private, 이름 맹글링)
- 추상 클래스(`ABC`, `@abstractmethod`), 정적 멤버(`@staticmethod`, 클래스 변수), dunder 메소드, 데코레이터를 함께 정리했다.

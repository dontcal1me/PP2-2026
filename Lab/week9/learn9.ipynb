# 제9장 GUI 프로그래밍 (tkinter)


## 1. Tkinter 시작하기
가장 기본적인 윈도우 생성 및 레이블(Label) 위젯 배치 방법입니다.

```python
from tkinter import *

window = Tk() # 루트 윈도우 생성
label = Label(window, text="Hello tkinter") # 레이블 위젯 생성
label.pack() # 레이블 위젯을 윈도우에 배치 (압축 배치 관리자)

window.mainloop() # 윈도우가 사용자 동작을 대기 (이벤트 루프)
```

## 2. 기본 위젯 (버튼, 엔트리)
버튼(Button)과 텍스트를 입력받는 엔트리(Entry) 위젯의 생성 방법입니다.

```python
from tkinter import *

window = Tk()

# 버튼 위젯
button = Button(window, text="클릭하세요!", bg="yellow", fg="blue", width=80, height=2)
button.pack()

# 엔트리 위젯
entry = Entry(window, fg="black", bg="yellow", width=80)
entry.pack()

window.mainloop()
```

## 3. 배치 관리자 (Geometry Managers)
tkinter는 위젯을 배치하기 위해 `pack`, `grid`, `place` 세 가지 배치 관리자를 제공합니다.

```python
from tkinter import *

window = Tk()
window.geometry("300x200") # 윈도우 크기 설정

# 1. pack: 차곡차곡 쌓아 올리는 방식
# Button(window, text="박스 #1", bg="red").pack(side=LEFT)

# 2. grid: 격자(테이블) 형태로 행(row)과 열(column)을 지정하여 배치
b1 = Button(window, text="격자(0,0)", bg="red", fg="white")
b2 = Button(window, text="격자(0,1)", bg="green", fg="white")
b1.grid(row=0, column=0)
b2.grid(row=0, column=1)

# 3. place: 지정된 절대 좌표(x, y)에 배치
# Button(window, text="절대좌표", bg="orange").place(x=40, y=60)

window.mainloop()
```

## 4. 버튼 이벤트 처리 (카운터 만들기)
버튼을 눌렀을 때 특정 함수(이벤트 핸들러)가 실행되도록 `command` 옵션을 사용합니다.

```python
from tkinter import *

window = Tk()
counter = 0

def clicked():
    global counter
    counter += 1
    label['text'] = '버튼 클릭 횟수: ' + str(counter)

label = Label(window, text="아직 눌려지지 않음")
label.pack()

# command 매개변수에 콜백 함수(clicked) 연결
button = Button(window, text="증가", command=clicked)
button.pack()

window.mainloop()
```

## 5. 실습: 온도 변환기
Entry에서 값을 읽어와 계산한 뒤, 다른 Entry에 결과를 출력하는 프로그램입니다.

```python
from tkinter import *

def process():
    tf = float(e1.get()) # 화씨 온도 읽기
    tc = (tf-32.0)*5.0/9.0 # 섭씨로 변환
    e2.delete(0, END) # 기존 텍스트 지우기
    e2.insert(0, str(tc)) # 변환된 값 삽입

window = Tk()

Label(window, text="화씨").grid(row=0, column=0)
Label(window, text="섭씨").grid(row=1, column=0)

e1 = Entry(window)
e2 = Entry(window)
e1.grid(row=0, column=1)
e2.grid(row=1, column=1)

Button(window, text="화씨->섭씨", command=process).grid(row=2, column=1)

window.mainloop()
```

## 6. 실습: 계산기 UI
반복문과 리스트를 활용하여 `grid` 배치 관리자로 계산기 버튼들을 효율적으로 배치합니다.

```python
from tkinter import *

window = Tk()
window.title("My Calculator")
display = Entry(window, width=33, bg="yellow")
display.grid(row=0, column=0, columnspan=5)

button_list = [
    '7', '8', '9', '/', 'C',
    '4', '5', '6', '*', ' ',
    '1', '2', '3', '-', ' ',
    '0', '.', '=', '+', ' '
]

row_index = 1
col_index = 0

for button_text in button_list:
    Button(window, text=button_text, width=5).grid(row=row_index, column=col_index)
    col_index += 1
    if col_index > 4:
        row_index += 1
        col_index = 0

window.mainloop()
```

## 7. 캔버스(Canvas)에 그림 그리기
선, 사각형, 타원, 다각형 등의 도형을 그리고 색상 및 텍스트를 지정할 수 있습니다.

```python
from tkinter import *

window = Tk()
w = Canvas(window, width=300, height=200)
w.pack()

# 사각형 그리기 (x1, y1, x2, y2)
w.create_rectangle(50, 25, 200, 100, fill="blue")

# 텍스트 그리기
w.create_text(150, 150, fill="darkblue", font="Times 15 italic bold", text="This is Canvas")

window.mainloop()
```

## 8. 마우스 및 키보드 이벤트 처리 (`bind`)
위젯에 이벤트를 바인딩하여 마우스 클릭, 키보드 입력 등을 처리할 수 있습니다.

```python
from tkinter import *

window = Tk()
window.geometry("300x200")

label = Label(window, text="창 안을 클릭하세요")
label.pack()

def callback(event):
    label['text'] = f"{event.x}, {event.y} 에서 클릭!"
    print(event.x, event.y, "에서 마우스 클릭 이벤트 발생")

# <Button-1>은 마우스 왼쪽 버튼 클릭을 의미합니다.
window.bind("<Button-1>", callback)

window.mainloop()
```

```python
import os
import random
from tkinter import *

# 이미지가 있는 폴더 (노트북 실행 위치와 상관없이 찾도록 고정)
IMG_DIR = r"C:\Users\user\Desktop\Repo-202611831\PP2-2026\Lab\week9"

window = Tk()
window.title("가위바위보")

Label(window, text="선택하세요", font=("Helvetica", 16)).pack()
frame = Frame(window)
frame.pack()

# master=window: 이 창에 이미지를 등록 (노트북에서 Tk()를 여러 번 만들 때 필요)
rock_image = PhotoImage(master=window, file=os.path.join(IMG_DIR, "rock.png"))
paper_image = PhotoImage(master=window, file=os.path.join(IMG_DIR, "paper.png"))
scissors_image = PhotoImage(master=window, file=os.path.join(IMG_DIR, "scissors.png"))

result_label = Label(window, text="", font=("Helvetica", 14))
result_label.pack(pady=5)

def decide(user):
    computer = random.choice(["가위", "바위", "보"])
    if user == computer:
        result = "비겼습니다"
    elif (user, computer) in [("가위", "보"), ("바위", "가위"), ("보", "바위")]:
        result = "이겼습니다!"
    else:
        result = "졌습니다"
    result_label["text"] = f"나: {user} / 컴퓨터: {computer} → {result}"

def pass_s():
    decide("가위")

def pass_r():
    decide("바위")

def pass_p():
    decide("보")

Button(frame, image=scissors_image, command=pass_s).grid(row=0, column=0)
Button(frame, image=rock_image, command=pass_r).grid(row=0, column=1)
Button(frame, image=paper_image, command=pass_p).grid(row=0, column=2)

window.mainloop()
```
